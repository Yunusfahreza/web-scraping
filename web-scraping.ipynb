```python
import requests
from bs4 import BeautifulSoup
```

```python
# Tes Scrapping
# Base URL
base_url = 'https://serverkorea.indksn88.net/desktop/home'

def get_cleaned_text_from_url(url):
    # Kirim permintaan GET ke URL
    response = requests.get(url)

    # Dapatkan konten HTML
    html_content = response.text

    # Parsing konten HTML menggunakan BeautifulSoup
    soup = BeautifulSoup(html_content, 'html.parser')

    # Hapus elemen script dan style
    for script_or_style in soup(['script', 'style']):
        script_or_style.decompose()

    # Ekstrak teks dari konten HTML
    text = soup.get_text(separator=' ')

    # Pisahkan menjadi baris dan hapus spasi di awal/akhir setiap baris
    lines = (line.strip() for line in text.splitlines())

    # Pisahkan multi-headlines menjadi satu baris
    chunks = (phrase.strip() for line in lines for phrase in line.split("  "))

    # Hapus baris kosong
    text = '\n'.join(chunk for chunk in chunks if chunk)

    return text
```

```python
from Sastrawi.StopWordRemover.StopWordRemoverFactory import StopWordRemoverFactory
import re
import string
from Sastrawi.Stemmer.StemmerFactory import StemmerFactory
from nltk.tokenize import word_tokenize
from nltk.probability import FreqDist
from Sastrawi.Stemmer.StemmerFactory import StemmerFactory
import nltk
import matplotlib.pyplot as plt

nltk.download('punkt')
```

```python
# Fungsi untuk melakukan case folding dan pembersihan teks
def clean_text(text):
    # Ubah teks menjadi huruf kecil (case folding)
    text = text.lower()

    # Hapus angka
    text = re.sub(r'\d+', '', text)

    # Hapus tanda baca
    text = text.translate(str.maketrans('', '', string.punctuation))

    # Hapus spasi di awal dan akhir
    text = text.strip()

    return text


# Fungsi untuk menghapus stopwords dari teks
def remove_stopwords(text):
    stop_factory = StopWordRemoverFactory()
    more_stopword = ['dengan', 'ia', 'bahwa', 'oleh', 'kamu', 'hari', 'ini']
    stopwords = stop_factory.get_stop_words() + more_stopword

    # Buat string tanpa stopwords
    cleaned_text = ' '.join(word for word in text.split() if word not in stopwords)

    return cleaned_text
```

```python
# Fungsi untuk menghapus stopwords dari teks
def remove_stopwords(text):
    stop_factory = StopWordRemoverFactory()
    more_stopword = ['dengan', 'ia', 'bahwa', 'oleh', 'kamu', 'hari', 'ini']
    stopwords = stop_factory.get_stop_words() + more_stopword

    # Buat string tanpa stopwords
    cleaned_text = ' '.join(word for word in text.split() if word not in stopwords)

    return cleaned_text
```

```python
# URL yang ingin di-scrape
url = 'https://serverkorea.indksn88.net/desktop/home'

# Dapatkan dan bersihkan teks dari URL
cleaned_text = get_cleaned_text_from_url(url)

# Lakukan case folding dan pembersihan teks
cleaned_text = clean_text(cleaned_text)

# Hapus stopwords dari teks yang telah dibersihkan
final_text = remove_stopwords(cleaned_text)

# Cetak teks yang sudah dibersihkan dari stopwords
final_text
```

```python
# Tokenisasi dan frekuensi kemuculan

tokens = nltk.tokenize.word_tokenize(final_text)
kemunculan = nltk.FreqDist(tokens)
kemunculan.most_common(15)
```

```python
# Stemming

factory = StemmerFactory()
stemmer = factory.create_stemmer()
stemmed_text = stemmer.stem(final_text)
stemmed_text
```

```python
# Tokenisasi dan frekuensi kemuculan

tokens = nltk.tokenize.word_tokenize(stemmed_text)
kemunculan = nltk.FreqDist(tokens)
kemunculan.most_common(15)
```
